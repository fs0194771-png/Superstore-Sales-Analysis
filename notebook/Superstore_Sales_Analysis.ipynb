# Exploratory Data Analysis (EDA)

## Overall Business Performance 

```python
print("Total Sales ",df['sales'].sum())
total_gains = df.loc[df["profit"] > 0, "profit"].sum()
total_losses = df.loc[df["profit"] < 0, "profit"].sum()
total_profit_margin =(df["profit"].sum()/df['sales'].sum())*100
print("Total Gains:", total_gains)
print("Total Losses:", total_losses)
print("Net Profit:", df["profit"].sum())
print("total_profit_margin:", total_profit_margin)
print("total quantity",df['quantity'].sum())
print("total order",df['order_id'].nunique())
print("total customer",df['customer_id'].nunique())
print("Avg discount",df['discount'].mean()*100)
```

## Category Performance

```python
category_profit = df.groupby("category").agg(
    Sales=("sales", "sum"),
    Profit=("profit", "sum"),
    Quantity=("quantity", "sum")
)

category_profit["Profit Margin %"] = (
    category_profit["Profit"] / category_profit["Sales"]
) * 100

category_profit = category_profit.sort_values(
    "Profit", ascending=False
)

category_profit
```

##  Sub-category Performance

```python
subcategory_profit = df.groupby("sub_category").agg(
    Sales=("sales", "sum"),
    Profit=("profit", "sum"),
    Quantity=("quantity", "sum")
)

subcategory_profit["Profit Margin"] = (
    subcategory_profit["Profit"] / subcategory_profit["Sales"]
) * 100

subcategory_profit.sort_values("Profit")
```

## Discount Impact on Profitability

# Data Understanding Insights:
The dataset contains 9,994 records and 21 columns covering sales transactions from 2019 to 2022. It includes customer, product, geographic, sales, discount, profit, and shipping information. The dataset contains 3 categories, 16 sub-categories, 3 customer segments, and 4 regions.

# Data Cleaning Insights:
No missing values or duplicate rows were identified. Date columns were converted from strings to datetime format, and column names were standardized using lowercase and underscore notation. Data types and key numeric ranges were validated before proceeding to data wrangling and EDA
# Business Insights
1. Overall Business Performance
The business generated $5.76M in total sales, but recorded a net loss of $90.32K, resulting in an overall profit margin of -1.57%.
Total gains amounted to $348.50K, while total losses reached $438.82K, indicating that losses outweighed gains.
The average discount across all transactions was relatively high at 28.74%, suggesting that discounting may be a major factor affecting profitability.
2. Category Performance
Technology generated the highest profit at $64.55K, with a profit margin of 2.61%.
Office Supplies generated $44.25K in profit and achieved the highest profit margin of 4.73%.
Furniture was the main source of losses, generating $2.36M in sales but losing $199.12K, with a negative profit margin of -8.45%.
This shows that high sales volume does not necessarily translate into profitability, particularly in the Furniture category.
3. Sub-category Performance
All four Furniture sub-categories were loss-making:
Tables: -$50.65K profit, -8.73% margin.
Chairs: -$51.13K profit, -8.30% margin.
Furnishings: -$49.25K profit, -8.47% margin.
Bookcases: -$48.09K profit, -8.30% margin.
The losses are therefore not isolated to one Furniture sub-category, but represent a broader profitability problem within Furniture.
4. Discount and Profitability
Profitability decreases sharply as discounts increase.
Transactions with 0% discount generated $250.18K profit and a 14.70% profit margin.
Profit became negative starting at a 30% discount, reaching a loss of $106.10K at an 80% discount.
The correlation between discount and profit was -0.46, indicating a moderate negative relationship.
More importantly, the correlation between discount and profit margin was -0.92, showing a very strong negative relationship between discounting and profitability.
5. Discount Impact Across Categories
Furniture becomes loss-making even at relatively moderate discount levels, with its profit margin becoming negative at 15% discount.
Technology also becomes loss-making at higher discount levels, while Office Supplies remains profitable for longer.
At 80% discount, all three categories are strongly loss-making.
This indicates that discounting has different effects across categories, and a single discount strategy should not be applied to all products.
6. Yearly Performance
Sales remained relatively stable across the four years, ranging from approximately $1.37M to $1.49M.
However, profitability varied significantly.
2020 was the worst-performing year, with a loss of $38.82K and a profit margin of -2.61%.
The other years also remained unprofitable, but their losses were lower.
Since average discounts were relatively similar across years, the 2020 decline in profitability cannot be explained by the average discount alone.
7. Year × Category Performance
Furniture remained loss-making in every year, confirming that its profitability problem is structural rather than limited to a specific year.
Furniture's worst performance occurred in 2020, when it recorded a loss of $55.63K and a profit margin of -9.19%.
Technology remained profitable every year, although its profitability declined considerably in 2020.
Office Supplies also remained profitable throughout the period.
8. Regional Performance
Furniture generated losses across all four regions, confirming that the issue is not limited to one geographic area.
The largest Furniture losses occurred in:
East: -$63.32K
West: -$60.88K
Central: -$47.02K
South: -$27.90K
This further supports the conclusion that Furniture has a broad profitability problem rather than a purely regional one.
9. City Performance
Several cities generated significant overall losses, with Ohio City, Texas City, Utah City, North Carolina City, and Nevada City among the largest loss-making cities.
The city-level analysis indicates that profitability problems are concentrated in certain locations and should be investigated alongside product/category performance.
10. Customer Performance
A small group of customers generated relatively high sales but significant losses.
The highest loss-making customers recorded losses ranging from approximately $1.82K to $3.03K.
This indicates that high-value customers are not necessarily profitable customers, particularly when large discounts or low-margin products are involved.
11. Customer Segment Performance
All three customer segments were unprofitable:
Consumer: -$38.30K
Corporate: -$25.17K
Home Office: -$26.86K
Home Office had the worst profit margin at -2.66%, compared with -1.28% for Consumer and -1.43% for Corporate.
Therefore, profitability issues are present across all segments, with Home Office requiring particular attention.
12. Shipping Performance
Profitability did not show a clear relationship with shipping duration.
All shipping-duration groups remained loss-making, but there was no consistent pattern showing that longer shipping times caused higher losses.
Therefore, shipping time does not appear to be a major driver of the overall profitability problem.
13. Monthly Performance
The business remained loss-making across all months.
The strongest monthly results were observed around September, July, and March, where losses were relatively small.
The weakest months included August, June, January, and May.
However, there is no clear seasonal pattern strong enough to explain the overall profitability problem.

# Recommendations
1. Reassess the Discount Strategy

The company should reduce excessive discounting, particularly discounts of 30% or more, because profitability becomes negative at this level.

Discounts should be based on:

Product profitability
Category
Customer value
Demand
Existing margin
2. Review the Furniture Category

Furniture should be the highest priority for profitability improvement because it is responsible for the largest overall losses.

The company should investigate:

Product costs
Pricing
Supplier costs
Discount levels
Product-level margins
Potentially unprofitable products
3. Apply Category-Specific Discount Policies

The same discount percentage should not be applied across all categories.

For example:

Technology and Office Supplies can support different discount levels.
Furniture requires much stricter discount controls because it becomes unprofitable at relatively low discount levels.
4. Review Loss-Making Furniture Sub-categories

Since Chairs, Tables, Furnishings, and Bookcases are all loss-making, the company should review the pricing and cost structure of the entire Furniture category rather than focusing on a single sub-category.

5. Investigate High-Loss Products

Products with high sales but strongly negative profit margins should be reviewed individually.

Management should consider:

Increasing prices
Reducing discounts
Negotiating supplier costs
Discontinuing consistently unprofitable products
6. Investigate the 2020 Profitability Decline

Since sales were relatively stable while profitability dropped significantly in 2020, the company should investigate what operational or pricing factors caused the additional losses during that year.

7. Focus on Profitable Categories

Technology and Office Supplies should be maintained and potentially expanded because they generate positive profits and margins.

However, growth should still be evaluated based on profitability rather than sales volume alone.

8. Monitor Customer Profitability

The company should evaluate customers based on profit generated, not just revenue.

High-revenue customers generating repeated losses should be reviewed for:

Excessive discounts
Product mix
Order size
Pricing conditions
9. Investigate High-Loss Locations

Cities and regions with consistently negative profitability should be investigated to determine whether the issue comes from:

Product mix
Discounting
Pricing
Customer behavior
Operating costs
10. Do Not Prioritize Shipping Changes as the Main Solution

Since shipping duration did not show a clear relationship with profitability, reducing shipping time should not be considered the primary solution to the company's financial losses.

The analysis suggests that pricing, discounting, and product profitability are much more important areas to address.

# Conclusion

The analysis shows that the company's main challenge is not generating sales, but converting sales into profit. Despite generating $5.76M in sales, the business recorded a net loss of $90.32K and an overall profit margin of -1.57%. Furniture is the primary source of losses, while Technology and Office Supplies remain profitable. The analysis also revealed a strong negative relationship between discounting and profitability, with profit margins declining sharply as discounts increase. The profitability problem is present across different years, regions, cities, and customer segments, while shipping duration does not appear to be a major driver. Therefore, the main opportunities for improvement are controlling excessive discounts, restructuring the Furniture category, reviewing loss-making products, and focusing growth on profitable products and categories rather than sales volume alone.
